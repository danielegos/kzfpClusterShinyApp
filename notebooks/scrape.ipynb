```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager


def get_pli_selenium(ensembl_gene_id, dataset="gnomad_r4", headless=True):
    url = f"https://gnomad.broadinstitute.org/gene/{ensembl_gene_id}?dataset={dataset}"

    # Set up Chrome options
    options = webdriver.ChromeOptions()
    if headless:
        # new headless mode for recent Chrome
        options.add_argument("--headless=new")
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")

    # Start driver (webdriver-manager auto-downloads the right chromedriver)
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()),
                              options=options)

    try:
        driver.get(url)

        wait = WebDriverWait(driver, 20)

        # 1. Wait for the "pLI" label to appear somewhere in the page
        pli_label_elem = wait.until(
            EC.presence_of_element_located(
                (By.XPATH, "//*[normalize-space(text())='pLI']")
            )
        )

        # 2. From that label, grab the next <span> with "TooltipHint" in its class
        pli_span = pli_label_elem.find_element(
            By.XPATH,
            "following::span[contains(@class, 'TooltipHint')][1]"
        )

        pli_value = pli_span.text.strip()
        return pli_value

    finally:
        driver.quit()


if __name__ == "__main__":
    gene_id = "ENSG00000196453"
    pli = get_pli_selenium(gene_id)
    print(f"pLI for {gene_id}: {pli}")
```

```python
import requests
import json
from bs4 import BeautifulSoup

def get_pli(ensembl_gene_id: str, dataset: str = "gnomad_r4"):
    url = f"https://gnomad.broadinstitute.org/gene/{ensembl_gene_id}?dataset={dataset}"
    headers = {
        "User-Agent": (
            "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
            "(KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36"
        )
    }

    # 1. Download the page
    resp = requests.get(url, headers=headers)
    resp.raise_for_status()

    # 2. Parse HTML and find the JSON payload
    soup = BeautifulSoup(resp.text, "html.parser")
    script_tag = soup.find("script", id="__NEXT_DATA__")
    if script_tag is None or not script_tag.string:
        raise RuntimeError("Could not find __NEXT_DATA__ script in the page HTML")

    data = json.loads(script_tag.string)

    # 3. Recursively search for the pLI key
    def find_pli(node):
        if isinstance(node, dict):
            if "pLI" in node:
                return node["pLI"]
            for v in node.values():
                result = find_pli(v)
                if result is not None:
                    return result
        elif isinstance(node, list):
            for item in node:
                result = find_pli(item)
                if result is not None:
                    return result
        return None

    pli_value = find_pli(data)
    if pli_value is None:
        raise RuntimeError("pLI not found in __NEXT_DATA__ JSON")

    return pli_value


if __name__ == "__main__":
    gene_id = "ENSG00000196453"
    pli = get_pli(gene_id)
    print(f"pLI for {gene_id}: {pli}")
```

```python
import pandas as pd

# gnomAD v4.1 gene/transcript constraint metrics table
CONSTRAINT_URL = (
    "https://storage.googleapis.com/gcp-public-data--gnomad/"
    "release/4.1/constraint/gnomad.v4.1.constraint_metrics.tsv"
)

def get_pli_from_constraint(ensembl_gene_id: str, constraint_url: str = CONSTRAINT_URL):
    """
    Return the pLI value for a given Ensembl gene ID using the
    gnomAD v4.1 constraint metrics table.
    """
    # Read directly from the URL (pandas will stream it)
    df = pd.read_csv(constraint_url, sep="\t")

    # Inspect available columns if you're curious:
    # print(df.columns.tolist())

    # Try to guess the pLI column name
    pli_col = None
    for cand in ["pLI", "_pli", "pli", 'lof.pLI']:
        if cand in df.columns:
            pli_col = cand
            break

    if pli_col is None:
        raise RuntimeError(
            "Could not find a pLI column in the constraint table. "
            f"Available columns: {df.columns.tolist()}"
        )

    # Filter by Ensembl gene ID
    gene_rows = df[df["gene_id"] == ensembl_gene_id]

    if gene_rows.empty:
        raise ValueError(f"Gene {ensembl_gene_id} not found in constraint table.")

    # There may be multiple transcripts per gene; take the canonical / first row
    pli_value = gene_rows.iloc[0][pli_col]

    return pli_value
```
